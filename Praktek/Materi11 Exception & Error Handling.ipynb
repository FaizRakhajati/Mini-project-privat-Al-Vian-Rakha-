## Exception & Error Handling

Exception & Error Handling adalah mekanisme di Python untuk mengontrol dan menangani runtime error agar aplikasi tidak mendadak crash, melainkan bisa merespons error tersebut dengan baik.

### Jenis-Jenis Exception

1. ZeroDivisionError : 

Terjadi saat membagi angka dengan nol (5 / 0).

2. ValueError : 

Nilai argumen tidak valid (misal: int("abc")).

3. TypeError : 

Operasi dilakukan pada tipe data yang tidak sesuai (misal: "2" + 5).

4. KeyError : 

Mengakses key dictionary yang tidak terdaftar.

5. IndexError : 

Mengakses indeks list yang melebihi batas.

6. FileNotFoundError : 

File yang ingin dibuka tidak ditemukan di sistem.

### Struktur Exception Dasar

1. `try` 

Digunakan untuk kode yang akan di coba / kode yang berpotensi error

2. `except` `Type Errornya` 

Digunakan untuk kode yang akan dieksekusi jika error sesuai dengan Type Error yang dicantumkan

3. `else` 

Digunakan untuk kode yang akan dieksekusi jika kode pada try tidak error (Opsional)

4. `finnaly` 

Selalu tereksekusi apapun kejadiannya (Opsional)

```python
try:
    angka = int(input("Masukan Angka"))
except ValueError:
    print("Input Wajib Menggunakan Angka")
else:
    print("Input Berhasil")
finally:
    print("Blok finally selalu tereksekusi")
```

### Multi Exception

```python
buah = ["Mangga","Pisang","Jambu"]

try :
    indeks = int(input("Masukan Indeks Buah yang ingin dihapus : "))
    buah.pop(indeks)
except ValueError : 
    print("Indeks Wajib Menggunakan Angka")
except IndexError :
    print("Indeks tidak boleh lebih dari 2 atau negatif")
finally :
    print(buah)
    print("Exception Selesai")
    
```

```python
buah = ["Mangga","Pisang","Jambu"]

try :
    indeks = int(input("Masukan Indeks Buah yang ingin dihapus : "))
    buah.pop(indeks)
except ValueError,IndexError : 
    print("Indeks Wajib Menggunakan Angka dan Indeks tidak boleh lebih dari 2 atau negatif")
finally :
    print(buah)
    print("Exception Selesai")
    
```
