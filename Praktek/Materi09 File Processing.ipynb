## File Processing

Pada program sederhana, Biasanya File Processing digunakan untuk Menyimpan dan Mengambil data agar data tidak hilang pada saat program selesai digunakan.

### Mode Akses File

Mode akses file digunakan ketika program akan membuka file

Macam Macam Mode Akses File :

1. `w` untuk membuat file baru (jika file ada, data file lama akan hilang)

2. `r` untuk membaca file (jika file tidak ada, akan error)

3. `a` untuk mengedit file (jika file tidak ada, akan membuat file baru)

dll.



### Membuka dan Menutup File

Cara paling aman membuka dan menutup file tanpa menggunakan library adalah dengan menggunakan sintaks `with`

sintaks dasar :

`with open (`nama file`,`Mode Akses`) as `variabel bebas` : `

```python
with open("FileMateri09.csv","w") as filebaru :
    pass
```

pada sintaks `open()`, terdapat parameter bernama `encoding`

pada paremeter ini, disarankan menggunakan `utf-8` karena UTF-8 adalah standar global saat ini

```python
with open("FileMateri09.csv","w", encoding = "utf-8") as filebaru :
    print("Prabs",file = filebaru)
```

### Menyimpan Data

```python
data = [
    ["Artificial Intelligence for Dummies",5],
    ["The Power of Kepepet",8],
    ["Hujan Kemarin",6],
    ["Data Structure and Algorithm 2nd Edition",10],
    ["Hundred Pages of Machine Learning",2]
]
```

```python
with open("FileMateri09.csv","w",encoding="utf-8") as ffile :
    for i in range(len(data)) :
        print(data[i][0] , data[i][1] , sep="," , file=ffile)
```

### Ekstraksi Data

```python
data = []
print(data)
```

Ekstraksi data dapat menggunakan `for`, Setiap iterasi akan memuat 1 baris menjadi 1 kalimat `string`

Tips :

1. gunakan `strip()` untuk menghilangkan `endline`/`\n`

2. lalu gunakan `split()` untuk memisahkan antar data

3. lalu jangan lupa untuk mengubah ke type variabel yang cocok untuk setiap data

```python
with open("FileMateri09.csv","r",encoding="utf-8") as ffile :
    for baris in ffile :
        # 1
        baris = baris.strip()
        
        # 2 
        buku , stok = baris.split(",")
        
        # 3
        stok = int(stok)
        
        data.append([buku,stok])

print(data)
```
